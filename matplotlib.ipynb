Let's talk about the customisation of graphs. What comes under the customisation?
1. A marker: it represents what we should represent the data points with. It might be a circle or a triangle, etc.
2. Colour: you don't need to write the entire colour name, like red or yellow. You just need to enter the first letter, such as R and Y and B. That's enough.
3. Linestyle: it represents the style of the line that a graph should represent the data points with. It might be dotted lines, dashed lines, or a solid line, etc.
4. Linewidth: it is the fourth customisation parameter. Line width represents it as 5, 6, 7, 8, etc., and it will increase the width of the line.
 If you find that any one of them is not found, then simply refer to its documentation. Not a big deal.

Let's dive into bar graph. We use bar graph when we have categorical data. Categorical data means when we say that the data belongs to that category or this category. It doesn't mean that it should be string type. For example, we have a real-world example like movies revenue under the release year of release against to the revenue. They both are categorical because that particular movie has released in a particular category here 2008. So it doesn't mean that they should not be numerics. They should be numerics even if the categorical data. Let's Plot bar chart. 

```python
academic_year = ["2023-2024","2024-2025","2025-2026"]
pass_percentage = ["78%","83%","81%"]
plt.bar(academic_year,pass_percentage,label = "student performance",color="g")
plt.title("studies record")
plt.xlabel("year")
plt.ylabel("percentage")
plt.legend()
plt.grid()
for i in range(len(academic_year)):#Here we have added the text over the bars by using a method called `text()` that is a special method in Python used to add content over the bars. It includes four parameters:
# - Location of x
# - Location of y
# - What content to be displayed over the bar
# - Location and exact text to be present
    plt.text(academic_year[i],pass_percentage[i],pass_percentage[i],ha="center")
plt.show()
```

```python
# Now we will draw the bar graph horizontally. 
plt.barh(academic_year,pass_percentage,label = "student performance",color="r")
plt.text(pass_percentage[1],academic_year[1],pass_percentage[1],ha="center")
plt.show()
```

```python
import numpy as np
years = [2020,2021,2022,2023]
oscar_revenue = [101,102,103,104]
non_o_r = [89,67,98,90]
x=np.arange(len(years))
width = 0.4
plt.bar(x-width/2,oscar_revenue,width=width,color ="r")
plt.bar(x+width/2,non_o_r,width=width,color ="g")
plt.xticks(x,years)
# plt.xlim(0,10)
plt.ylim(0,150)
for i in range(len(years)):
    plt.text(years[i],non_o_r[i],str(non_o_r),ha="center")
plt.savefig("bar.png")
plt.show()
```

scatter plot used for finding the relation between two variables
1.corelation
2.outliers
3.clusters . it can be plot using 
plt.scatter(x,y)



```python
age = [50,55,60,65]
bp = [100,72,103,105]
colors = ["green" if i>100 else "red" for i in bp]
plt.scatter(age,bp,color=colors,s = bp,alpha=0.5)
for i in range(len(age)):
    plt.annotate(str(bp[i]),(age[i],bp[i]),(age[i]+1,bp[i]+1))
plt.colorbar(label="bp")
plt.show()
```
